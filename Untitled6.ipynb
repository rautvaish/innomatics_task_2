```python
#Problem 1: Unique Words in a Sentence
# Input sentence
sentence = "python is easy and python is powerful"

# Split the sentence into words based on spaces
words = sentence.split()

# Convert the list to a set to filter out duplicates
unique_words = set(words)

# Display results
print(f"Unique words count: {len(unique_words)}")
print(f"Unique words: {unique_words}")
```

```python
#Problem 2: Highest Salary from Employee Data
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000 }

# Find the employee with the maximum salary

highest_paid_employee = max(employees, key=employees.get)
salary = employees[highest_paid_employee]

print(f"Highest Salary: {highest_paid_employee} - {salary}")
```

```python
#Problem 3: Find Maximum and Minimum Values

numbers = [45, 22, 89, 10, 66]

# Initializing with the first element of the list
max_val = numbers[0]
min_val = numbers[0]

# Loop through the list tocompare values
for num in numbers:
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num

print(f"List: {numbers}")
print(f"Max: {max_val}")
print(f"Min: {min_val}")
```

```python
#Problem 4: Count Products Above a Price Threshold
prices = [450,1200, 899, 1500, 300]
threshold = 1000
count = 0

# Check each price against the threshold
for price in prices:
    if price > threshold:
        count += 1

print(f"Products above 1000: {count}")
```

```python
#Problem 5: Calculate Attendance Percentage
attendance = ["P", "P", "A", "P", "P"]

# Count 'P' for Present
present_count =attendance.count("P")
total_days = len(attendance)

# Calculate percentage
attendance_percentage = (present_count / total_days)* 100

print(f"Attendance Percentage: {attendance_percentage}")
```

```python
#Problem 6:Remove Duplicate Phone Numbers
phone_numbers = [9876543210, 9123456789,9876543210]

# Sets do not allow duplicate entries
unique_numbers = set(phone_numbers)

print(f"Unique phone numbers: {unique_numbers}")
```

```python
#Problem 7:Count Character Frequency
input_string ="pythonp"
frequency = {}

# Iterate through each character
for char in input_string:
    # If char is already in dict, increment,otherwise, set to 1
    frequency[char] =frequency.get(char, 0) + 1

print(frequency)
```

```python
#Problem 8:Convert List to Tuple

my_list = [10, 20, 30]

# Simple conversion using the tuple constructor
my_tuple =tuple(my_list)

print(f"Tuple: {my_tuple}")
```

```python
#Problem 9: Check if a Key Exists in a Dictionary

employee_data = {"Name": "Ravi", "ID": 101, "Dept": "Data Science"}
key_to_check = "Name"

# Using the 'in' keyword for membership testing
if key_to_check in employee_data:
    print("Employee exists")
else:
    print("Employee not found")
```
